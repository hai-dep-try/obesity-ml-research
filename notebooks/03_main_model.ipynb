# 03 — Main model (CatBoost)

CatBoost với hyperparam Table 3 paper (`lr=0.05`, `depth=8`, `l2_leaf_reg=5`, `iterations=300`).
Ưu tiên `python -m src.train` từ repo root.

```python
from pathlib import Path
import sys

import pandas as pd
from IPython.display import Image, display

REPO_ROOT = Path.cwd()
if not (REPO_ROOT / "src").exists():
    REPO_ROOT = REPO_ROOT.parent
sys.path.insert(0, str(REPO_ROOT))

main_csv = REPO_ROOT / "results" / "main_results.csv"
all_csv = REPO_ROOT / "results" / "all_models_results.csv"

if not main_csv.exists():
    from src.train import train

    train(tune_baselines=True, catboost_full_grid=False)

display(pd.read_csv(main_csv))
display(pd.read_csv(all_csv))
```

```python
fig_dir = REPO_ROOT / "results" / "figures"
for name in ["catboost_confusion_matrix.png", "catboost_feature_importance.png"]:
    path = fig_dir / name
    print(path)
    if path.exists():
        display(Image(filename=str(path)))
```

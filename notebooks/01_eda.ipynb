Khám phá dataset Obesity Levels (nhãn `NObeyesdad`).

Chạy từ repo root: đảm bảo đã `python -m src.download_data`.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

import matplotlib.pyplot as plt
import seaborn as sns

from src.preprocess import TARGET_COL, load_raw, preprocess

df = load_raw()
print(df.shape)
df.head()
```

```python
print(df[TARGET_COL].value_counts())
df.describe(include="all").T
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
order = df[TARGET_COL].value_counts().index
sns.countplot(data=df, y=TARGET_COL, order=order, ax=ax)
ax.set_title("Phân bố lớp NObeyesdad")
fig.tight_layout()
out = ROOT / "results/figures/eda_class_distribution.png"
out.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(out, dpi=150)
plt.show()
print("Saved", out)
```
